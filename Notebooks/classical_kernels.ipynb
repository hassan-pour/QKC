## 1.Initializing

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold
from sklearn.svm import SVC

import sys

sys.path.append(".\\..")
from src.metrics import get_scores, scores_structure

# Load dataset
data = load_breast_cancer()


kf = StratifiedKFold(5, random_state=0, shuffle=True)

X = data.data[:, :]
y = data.target
```

## 2.Linear Model

```python
linear_scores_avg = scores_structure.copy()
    

for train_ind, test_ind in kf.split(X, y):
    X_train, X_test = X[train_ind], X[test_ind]
    y_train, y_test = y[train_ind], y[test_ind]

    # Model
    linear = SVC(kernel="linear")
    linear.fit(X_train, y_train)
    scores = get_scores(y_test, linear.predict(X_test))
    for key in scores.keys():
        linear_scores_avg[key] += scores[key] / 5


```

## 3.RBF Model

```python
rbf_scores_avg = scores_structure.copy()
    

for train_ind, test_ind in kf.split(X, y):
    X_train, X_test = X[train_ind], X[test_ind]
    y_train, y_test = y[train_ind], y[test_ind]

    # Model
    rbf = SVC(kernel="rbf")
    rbf.fit(X_train, y_train)
    scores = get_scores(y_test, rbf.predict(X_test))
    for key in scores.keys():
        rbf_scores_avg[key] += scores[key] / 5

    
```

## 4.Polynomial Model

```python
poly_scores_avg = scores_structure.copy()
    

for train_ind, test_ind in kf.split(X, y):
    X_train, X_test = X[train_ind], X[test_ind]
    y_train, y_test = y[train_ind], y[test_ind]

    # Model
    poly = SVC(kernel="poly")
    poly.fit(X_train, y_train)
    scores = get_scores(y_test, poly.predict(X_test))
    for key in scores.keys():
        poly_scores_avg[key] += scores[key] / 5
```

## 5.Sigmoid Model

```python
sigmoid_scores_avg = scores_structure.copy()
    

for train_ind, test_ind in kf.split(X, y):
    X_train, X_test = X[train_ind], X[test_ind]
    y_train, y_test = y[train_ind], y[test_ind]

    # Model
    sigmoid = SVC(kernel="sigmoid")
    sigmoid.fit(X_train, y_train)
    scores = get_scores(y_test, sigmoid.predict(X_test))
    for key in scores.keys():
        sigmoid_scores_avg[key] += scores[key] / 5
    
```

## 6.Results

```python
print(f"Linear Scores average: {linear_scores_avg}")
print(f"RBF Scores average: {rbf_scores_avg}")
print(f"Polynomial Scores average: {poly_scores_avg}")
print(f"Sigmoid Scores average: {sigmoid_scores_avg}")
```
